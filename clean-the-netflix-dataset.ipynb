```python
import pandas as pd

df = pd.read_csv("/kaggle/input/datasets/shivamb/netflix-shows/netflix_titles.csv")
df_raw = df.copy()   # salinan data mentah untuk perbandingan
```

```python
kosong = df.isnull().sum()
persen = (df.isnull().mean() * 100).round(2)
pd.DataFrame({"jumlah_kosong": kosong, "persen": persen}).sort_values("jumlah_kosong", ascending=False)
```

```python
df["rating"].value_counts(dropna=False)
```

```python
df[df["rating"].str.contains("min", na=False)][["show_id", "title", "rating", "duration"]]
```

```python
mask = df["rating"].str.contains("min", na=False)
df.loc[mask, "duration"] = df.loc[mask, "rating"]
df.loc[mask, "rating"] = None
```

```python
# Isi "Unknown" untuk kolom teks
for kolom in ["director", "cast", "country", "rating"]:
    df[kolom] = df[kolom].fillna("Unknown")

# Hapus baris tanpa tanggal
df = df.dropna(subset=["date_added"])

# Verifikasi
df.isnull().sum()
```

```python
print("Baris sebelum:", len(df_raw), "| sesudah:", len(df))
```

```python
df["duration_value"] = df["duration"].str.extract(r"(\d+)")[0].astype("Int64")
df["duration_unit"] = (df["duration"].str.extract(r"([A-Za-z]+)")[0]
                       .str.lower().str.rstrip("s"))
df[["title", "duration", "duration_value", "duration_unit"]].head()
```

```python
df["duration_unit"].value_counts()
df.groupby("type")["duration_unit"].value_counts()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"].str.strip(), format="%B %d, %Y")
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month
df[["date_added", "year_added", "month_added"]].head()
df["date_added"].dtype
```

```python
print("Duplikat baris:", df.duplicated().sum())
print("Duplikat show_id:", df["show_id"].duplicated().sum())
print("Duplikat title dan type:", df.duplicated(subset=["title", "type"]).sum())

for kolom in df.select_dtypes(include="object").columns:
    df[kolom] = df[kolom].str.strip()
```

```python
df.info()
print(df.isnull().sum().sum(), "data kosong tersisa")
df.to_csv("netflix_titles_cleaned.csv", index=False)
```

```python
df["country"] = (df["country"]
                 .str.strip(", ")
                 .str.replace(r"\s*,\s*", ", ", regex=True))
print(df["country"].str.contains(r"^,|,$").sum())   # harus 0
```

```python
df.to_csv("netflix_titles_cleaned.csv", index=False)
```

### Temuan penting

1. Tiga baris (film stand-up Louis C.K.) memiliki durasi yang salah masuk ke kolom rating
   ("66 min", "74 min", "84 min"). Nilai tersebut dipindahkan ke kolom duration, bukan dibuang.
2. Sepuluh baris yang dihapus (data date_added kosong) seluruhnya berasal dari serial TV,
   sehingga tidak ada film yang hilang.
3. Kolom director, cast, dan country diisi "Unknown" agar data tetap utuh dan tidak
   menggeser distribusi nilai.
4. Kategori rating NR dan UR dibiarkan apa adanya.

### Keterbatasan

Nilai "Unknown" pada director, cast, dan country berarti informasinya tidak tersedia,
bukan berarti memang tidak ada. Analisis yang memakai kolom tersebut perlu
mengecualikan atau memperlakukan nilai ini secara terpisah.
